```python
from pathlib import Path
import pandas as pd
import numpy as np

def find_repo_root():
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p/'requirements.txt').exists() and (p/'data').is_dir(): return p
    raise RuntimeError('Repository root not found. Run from inside the cloned repository.')
ROOT=find_repo_root()
print('Repository root:', ROOT)
```

```python
recovery=pd.read_csv(ROOT/'data/Figure1/Figure1_TitleOnlyRecovery_DERIVED_1453.csv')
assert len(recovery)==1453 and recovery.Broad_ID.nunique()==1453
counts=recovery.recovery_status.value_counts()
assert int(counts.get('RECOVERED',0))==192
assert int(counts.sum())==1453
print('Figure 1 title-only abstract recovery: PASS')
print(counts.to_string())
print('Note: bibliographic text and live API retrieval are intentionally excluded from the public SLIM repository.')
```
