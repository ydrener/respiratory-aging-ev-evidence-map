```python
from pathlib import Path
import pandas as pd
import numpy as np

def find_repo_root():
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p/'requirements.txt').exists() and (p/'data').is_dir(): return p
    raise RuntimeError('Repository root not found. Run from inside the cloned repository.')
ROOT=find_repo_root()
print('Repository root:', ROOT)
```

```python
recovery=pd.read_csv(ROOT/'data/Figure1/Figure1_TitleOnlyRecovery_DERIVED_1453.csv')
s=pd.read_csv(ROOT/'data/Figure1/Figure1_TitleOnlySafety_DERIVED.csv').set_index('Metric')['Count']
assert len(recovery)==1453
assert int(s['Title_only_total'])==1453
assert int(s['Recovered_abstracts'])==192
assert int(s['Article_Letter_audited'])==178
assert int(s['Article_Letter_not_relevant'])==178
assert int(s['High_signal_reviewed'])==19
assert int(s['Additional_advanced_to_full_text'])==0
print('Figure 1 title-only safety lane: PASS')
print(s.to_string())
```
