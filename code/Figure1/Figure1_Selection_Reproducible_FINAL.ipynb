# Figure 1 — Selection and full-text outcomes

```python
from pathlib import Path
import pandas as pd
import numpy as np

def find_repo_root():
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p/'requirements.txt').exists() and (p/'data').is_dir(): return p
    raise RuntimeError('Repository root not found. Run from inside the cloned repository.')
ROOT=find_repo_root()
print('Repository root:', ROOT)
```

```python
search=pd.read_csv(ROOT/'data/Figure1/Figure1_SearchCounts_DERIVED.csv').set_index('Metric')['Count']
corpus=pd.read_csv(ROOT/'data/Figure1/Figure1_Corpus_DERIVED_14366.csv')
ref28=pd.read_csv(ROOT/'data/Figure1/Figure1_Reference28_DERIVED.csv')
recovery=pd.read_csv(ROOT/'data/Figure1/Figure1_TitleOnlyRecovery_DERIVED_1453.csv')
safety=pd.read_csv(ROOT/'data/Figure1/Figure1_TitleOnlySafety_DERIVED.csv').set_index('Metric')['Count']
ft=pd.read_csv(ROOT/'data/Figure1/Figure1_FullTextOutcomes_DERIVED_51.csv')
dup=pd.read_csv(ROOT/'data/Figure1/Figure1_DuplicatePair_DERIVED.csv')
assert len(corpus)==14366 and int(corpus.has_abstract.sum())==12913 and int((corpus.has_abstract==0).sum())==1453
assert len(ref28)==28 and ref28['found'].astype(bool).all()
assert len(recovery)==1453 and int((recovery.recovery_status=='RECOVERED').sum())==192
assert int(safety['Additional_advanced_to_full_text'])==0
assert len(ft)==51
vc=ft.Final_eligibility.value_counts().to_dict(); assert vc=={'INCLUDE':37,'EXCLUDE':13,'NOT_RETRIEVED':1}, vc
assert ft.loc[ft.Broad_ID.eq('B05345'),'Final_eligibility'].iloc[0]=='NOT_RETRIEVED'
reasons=ft.loc[ft.Final_eligibility.eq('EXCLUDE'),'Primary_exclusion_reason'].value_counts().to_dict()
assert reasons=={'RESPIRATORY_AXIS_INSUFFICIENT':10,'AGING_SENESCENCE_AXIS_INSUFFICIENT':2,'EV_AXIS_INSUFFICIENT':1}
assert len(dup)==2 and set(dup.Broad_ID)=={'B02025','B02026'}
print('Figure 1 selection: PASS')
print('Full-text outcomes: 37 included / 13 excluded / 1 not retrieved')
```
