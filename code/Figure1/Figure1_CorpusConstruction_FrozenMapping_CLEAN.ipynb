# Figure 1 — Corpus construction from public derived mapping

```python
from pathlib import Path
import pandas as pd
import numpy as np

def find_repo_root():
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p/'requirements.txt').exists() and (p/'data').is_dir(): return p
    raise RuntimeError('Repository root not found. Run from inside the cloned repository.')
ROOT=find_repo_root()
print('Repository root:', ROOT)
```

```python
search=pd.read_csv(ROOT/'data/Figure1/Figure1_SearchCounts_DERIVED.csv').set_index('Metric')['Count']
corpus=pd.read_csv(ROOT/'data/Figure1/Figure1_Corpus_DERIVED_14366.csv')
assert int(search['Raw_total'])==27768
assert len(corpus)==14366 and corpus.Broad_ID.nunique()==14366
assert int(corpus.raw_record_count.sum())==27768
assert int(corpus.raw_record_count.sum()-len(corpus))==13402
assert int(corpus.has_abstract.sum())==12913
assert int((corpus.has_abstract==0).sum())==1453
assert (pd.to_numeric(corpus.raw_record_count)==pd.to_numeric(corpus.raw_record_count_recomputed)).all()
print('Figure 1 corpus construction: PASS')
print({'raw':27768,'deduplicated':14366,'duplicates_removed':13402,'title_abstract':12913,'title_only':1453})
```
