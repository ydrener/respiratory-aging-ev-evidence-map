# Figure 1 — Deduplication rule audit (reference)

```python
from pathlib import Path
import pandas as pd
import numpy as np

def find_repo_root():
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p/'requirements.txt').exists() and (p/'data').is_dir(): return p
    raise RuntimeError('Repository root not found. Run from inside the cloned repository.')
ROOT=find_repo_root()
print('Repository root:', ROOT)
```

```python
corpus=pd.read_csv(ROOT/'data/Figure1/Figure1_Corpus_DERIVED_14366.csv')
assert len(corpus)==14366
assert int(corpus.raw_record_count.sum())==27768
allowed={'SINGLETON','DOI','PMID','TITLE_EXACT_NORM','SPECIAL/TRANSITIVE'}
assert set(corpus.reconstructed_primary_support.dropna()) <= allowed
summary=corpus.reconstructed_primary_support.value_counts().rename_axis('support').reset_index(name='groups')
assert int(summary.groups.sum())==14366
assert int(corpus.raw_record_count.sum()-len(corpus))==13402
print('Figure 1 deduplication reference audit: PASS')
print(summary.to_string(index=False))
```
